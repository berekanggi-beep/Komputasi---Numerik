```python
# Tugas Komputasi Numerik - Metode Bisection
#  akar persamaan persamaan yang saya cari adalah: x³ - 2 = 0

#  akar yang saya definisikan fungsi persamaannya yaitu
def cari_nilai(s):
    return s**3 - 2

# dibutuhkan pencarian untuk saya Menentukan rentang awal
batas_awal = 1
batas_akhir = 2

# Tingkat ketelitian yang diharapkan
batas_salah = 0.001

# saya harus  Proses pencarian akar
while (batas_akhir - batas_awal) / 2 > batas_salah:
    titik_tengah = (batas_awal + batas_akhir) / 2

    if cari_nilai(batas_awal) * cari_nilai(titik_tengah) < 0:
        batas_akhir = titik_tengah
    else:
        batas_awal = titik_tengah

# membutuhkan cara pengerjaan untuk Menampilkan hasil dari pencarian akar
print("Hasil pencarian akar yang saya dapatkan yaitu  persamaan x³ - 2 = 0")
print("Akar dan titik tengahnya yaitu =", titik_tengah)






```

akar persamaan yang saya cari adalah x*3-2=0
setelah itu saya mendefinisikan fungsi persamaanya yaitu def cari nilai s dan returnya s**3-2 setelah itu ditentukanlah batas awal=1 dan batas akhir=2 dan batas salahnya adalah 0.001 selanjutnya saya harus membuat proses akarnya. disini yang di cari akarnya yaitu batas awal, batas akhir dan titik tengah titik awal+ batas akhir =2 sedangkan titik tengahnya = 1.259765625
