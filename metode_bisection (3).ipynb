```python
# === Program Metode Bagi Dua — Sistem Persamaan Nonlinier ===
# Mencari nilai x dan y yang memenuhi dua persamaan sekaligus

# Fungsi pertama: x² + y² - 4 = 0
def fungsi1(x, y):
    return x*x + y*y - 4

# Fungsi kedua: x - y - 1 = 0
def fungsi2(x, y):
    return x - y - 1

# Batas selang pencarian
bawah_x = 1.0
atas_x = 2.0
bawah_y = 0.0
atas_y = 1.0

# Batas ketelitian dan maksimal pengulangan
batas_error = 0.0001
maks_ulang = 50

# Tampilan judul tabel
print('=' * 60)
print(f"{'Ke-':<6} {'Xmin':<9} {'Xmax':<9} {'Ymin':<9} {'Ymax':<9} {'Xtengah':<10} {'Ytengah':<10}")
print('-' * 60)

hitungan = 0

# Proses iterasi
while ((atas_x - bawah_x)/2 > batas_error and
       (atas_y - bawah_y)/2 > batas_error and
       hitungan < maks_ulang):

    tengah_x = (bawah_x + atas_x) / 2
    tengah_y = (bawah_y + atas_y) / 2

    hitungan = hitungan + 1
    print(f"{hitungan:<6} {bawah_x:<9.5f} {atas_x:<9.5f} {bawah_y:<9.5f} {atas_y:<9.5f} {tengah_x:<10.5f} {tengah_y:<10.5f}")

    # Cek dan perkecil selang untuk x
    if fungsi1(bawah_x, tengah_y) * fungsi1(tengah_x, tengah_y) < 0:
        atas_x = tengah_x
    else:
        bawah_x = tengah_x

    # Cek dan perkecil selang untuk y
    if fungsi2(tengah_x, bawah_y) * fungsi2(tengah_x, tengah_y) < 0:
        atas_y = tengah_y
    else:
        bawah_y = tengah_y

# Hasil akhir
print('-' * 60)
nilai_x = (bawah_x + atas_x) / 2
nilai_y = (bawah_y + atas_y) / 2

print("\n Hasil Akhir:")
print(f"   x = {nilai_x:.6f}")
print(f"   y = {nilai_y:.6f}")
print(f"   Jumlah pengulangan = {hitungan}")

# Cek hasil
print("\n Cek nilai fungsi (harus mendekati nol):")
print(f"   f1 = {fungsi1(nilai_x, nilai_y):.8f}")
print(f"   f2 = {fungsi2(nilai_x, nilai_y):.8f}")
```

Setelah menjalankan program, hasil yang keluar adalah:
