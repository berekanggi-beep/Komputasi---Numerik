```python
# Matriks Augmentasi Sistem Persamaan Linear 3 Variabel
# Contoh:
# 2x + y - z = 8
# -3x - y + 2z = -11
# -2x + y + 2z = -3

matriks = [
    [2, 1, -1, 8],
    [-3, -1, 2, -11],
    [-2, 1, 2, -3]
]

print("=== Matriks Augmentasi ===")
for baris in matriks:
    print(baris)
```

```python
def eliminasi_gauss(matriks):
    # Salin matriks agar tidak mengubah yang asli
    A = [baris[:] for baris in matriks]
    n = len(A)

    # Langkah 1: Eliminasi ke bentuk segitiga atas
    for i in range(n):
        # Jadikan elemen utama = 1
        pembagi = A[i][i]
        for j in range(i, n + 1):
            A[i][j] /= pembagi

        # Hilangkan baris di bawahnya
        for k in range(i + 1, n):
            faktor = A[k][i]
            for j in range(i, n + 1):
                A[k][j] -= faktor * A[i][j]

    # Langkah 2: Substitusi mundur
    hasil = [0] * n
    for i in range(n - 1, -1, -1):
        hasil[i] = A[i][n]
        for j in range(i + 1, n):
            hasil[i] -= A[i][j] * hasil[j]

    return hasil
```

```python
# Panggil fungsi dan hitung
x, y, z = eliminasi_gauss(matriks)

# Tampilkan hasil
print("=== Hasil Penyelesaian SPL ===")
print(f"Nilai x = {x}")
print(f"Nilai y = {y}")
print(f"Nilai z = {z}")
```
