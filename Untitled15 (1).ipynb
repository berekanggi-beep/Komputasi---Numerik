```python
# Matriks Augmentasi Sistem Persamaan Linear 3 Variabel
# Contoh:
# 2x + y - z = 8
# -3x - y + 2z = -11
# -2x + y + 2z = -3

matriks = [
    [2, 1, -1, 8],
    [-3, -1, 2, -11],
    [-2, 1, 2, -3]
]

print("=== Matriks Augmentasi ===")
for baris in matriks:
    print(baris)
```

- Bagian ini mendefinisikan sistem persamaan linear 3 variabel ke dalam bentuk **matriks augmentasi**. Kolom pertama sampai ketiga adalah koefisien dari \(x, y, z\), sedangkan kolom keempat adalah nilai konstanta di ruas kanan.
- Matriks disimpan dalam bentuk daftar bertingkat di Python.
- Perulangan `for` digunakan untuk mencetak setiap baris matriks ke layar agar mudah dibaca.
- **Tujuan:** Menyiapkan data awal yang akan diselesaikan menggunakan metode Eliminasi Gauss.

```python
# Fungsi untuk menyelesaikan SPL menggunakan Metode Eliminasi Gauss
def eliminasi_gauss(matriks):
    # Membuat salinan matriks agar data asli tidak berubah
    A = [baris[:] for baris in matriks]
    n = len(A)

    # ==========================================
    # Langkah 1: Proses Eliminasi
    # Mengubah matriks menjadi bentuk segitiga atas
    # ==========================================
    for i in range(n):
        # Jadikan elemen pada diagonal utama bernilai 1
        pembagi = A[i][i]
        for j in range(i, n + 1):
            A[i][j] /= pembagi

        # Hilangkan elemen di bawah baris utama
        for k in range(i + 1, n):
            faktor = A[k][i]
            for j in range(i, n + 1):
                A[k][j] -= faktor * A[i][j]

    # ==========================================
    # Langkah 2: Substitusi Mundur
    # Mencari nilai variabel mulai dari baris terbawah
    # ==========================================
    hasil = [0] * n
    for i in range(n - 1, -1, -1):
        hasil[i] = A[i][n]
        for j in range(i + 1, n):
            hasil[i] -= A[i][j] * hasil[j]

    # Mengembalikan hasil x, y, z
    return hasil
```

> **Tujuan:** Mendefinisikan fungsi `eliminasi_gauss` untuk menghitung solusi sistem persamaan linear.

**Penjelasan bagian per bagian:**

- `def eliminasi_gauss(matriks):` → Mendefinisikan fungsi yang menerima matriks sebagai masukan.
- `A = [baris[:] for baris in matriks]` → Membuat salinan matriks agar nilai asli tidak berubah saat perhitungan.
- **Bagian Eliminasi:** Mengubah matriks menjadi bentuk segitiga atas — elemen di bawah diagonal utama bernilai nol, sehingga persamaan lebih mudah diselesaikan.
- **Bagian Substitusi Mundur:** Menghitung nilai variabel mulai dari baris paling bawah (z), lalu ke atas untuk mendapatkan nilai y dan x.
- `return hasil` → Mengembalikan hasil perhitungan berupa nilai \(x, y, z\).

```python
# Memanggil fungsi Eliminasi Gauss dan menampilkan hasil
x, y, z = eliminasi_gauss(matriks)

# Menampilkan hasil perhitungan
print("=== Hasil Penyelesaian SPL ===")
print(f"Nilai x = {x}")
print(f"Nilai y = {y}")
print(f"Nilai z = {z}")
```

> **Tujuan:** Menjalankan fungsi eliminasi_gauss dan menampilkan hasil penyelesaian.

**Penjelasan baris per baris:**

- Baris pertama memanggil fungsi `eliminasi_gauss` dengan matriks yang sudah didefinisikan di Sel [1]. Hasil perhitungannya langsung disimpan ke dalam variabel `x`, `y`, dan `z`.
- Perintah `print` menampilkan judul hasil, lalu mencetak nilai masing-masing variabel agar terlihat jelas di layar.
- **Hasil yang diperoleh:**
  - Nilai \(x = 2.0\)
  - Nilai \(y = 3.0\)
  - Nilai \(z = -1.0\)

Artinya, solusi dari sistem persamaan tersebut adalah **\(x = 2\), \(y = 3\), dan \(z = -1\)**
