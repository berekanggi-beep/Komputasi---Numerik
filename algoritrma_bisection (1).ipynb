pertama saya mencari akar persamman dan saya mendapat akar persamaan yaitu x**3-2=0 selanjutnya menentukan fungsi persamaan dan memilih rentang awal dimana fungsi persamaanya yaitu f(x) = **3-2 dan rentang awal 1 dan 2 batas ketelitian 0.001 lalu disini kita juga mencari akar dari jarak antara batas akhir

```python
# Tugas Komputasi Numerik - Metode Bisection
# Mencari akar persamaan: x³ - 2 = 0

# 1. Definisikan fungsi persamaan
def f(x):
    return x**3 - 2

# 2. Tentukan rentang awal dan ketelitian
a = 1    # batas bawah
b = 2    # batas atas
toleransi = 0.001

# 3. Cek awal: f(a) dan f(b) harus bertanda berbeda
if f(a) * f(b) >= 0:
    print("Pilih rentang lain, f(a) dan f(b) harus beda tanda")
else:
    # 4. Proses iterasi metode bisection
    while (b - a) / 2 > toleransi:
        c = (a + b) / 2  # titik tengah

        if f(c) == 0:
            a = c
            b = c
            break
        elif f(a) * f(c) < 0:
            b = c  # akar ada di kiri
        else:
            a = c  # akar ada di kanan

    # 5. Hasil akhir
    akar = (a + b) / 2
    print("Akar persamaan x³ - 2 = 0 adalah:", akar)
```
