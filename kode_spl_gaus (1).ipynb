```python
# === Matriks Augmentasi ===
matriks = [
    [2, 1, -1, 8],
    [-3, -1, 2, -11],
    [-2, 1, 2, -3]
]

for baris in matriks:
    print(baris)
```

> **Tujuan bagian ini:** Mendefinisikan sistem persamaan linear 3 variabel ke dalam bentuk matriks augmentasi.

- **Kolom pertama sampai ketiga** → adalah koefisien dari variabel **\(x, y, z\)**
- **Kolom keempat** → adalah nilai konstanta yang ada di ruas kanan setiap persamaan

Matriks ini disimpan dalam bentuk **daftar bertingkat** di Python — yaitu sebuah daftar yang berisi daftar-daftar lain di dalamnya.

Perintah `for baris in matriks` digunakan agar kita tidak perlu menulis `print(matriks[0])`, `print(matriks[1])`, dan seterusnya satu per satu — cukup satu perulangan, semua baris langsung tercetak otomatis ke layar.

```python
# === Fungsi untuk menyelesaikan SPL menggunakan Metode Eliminasi Gauss ===
def eliminasi_gauss(matriks):
    # Membuat salinan matriks agar data asli tidak berubah
    A = [baris[:] for baris in matriks]
    n = len(A)

    # ==========================================
    # Langkah 1: Proses Eliminasi
    # Mengubah matriks menjadi bentuk segitiga atas
    # ==========================================
    for i in range(n):
        # Jadikan elemen pada diagonal utama bernilai 1
        pembagi = A[i][i]
        for j in range(i, n + 1):
            A[i][j] = A[i][j] / pembagi

        # Hilangkan elemen di bawah baris utama
        for k in range(i + 1, n):
            faktor = A[k][i]
            for j in range(i, n + 1):
                A[k][j] = A[k][j] - faktor * A[i][j]

    # ==========================================
    # Langkah 2: Substitusi Mundur
    # Mencari nilai variabel dari baris terbawah
    # ==========================================
    hasil = [0] * n
    for i in range(n - 1, -1, -1):
        for j in range(i + 1, n):
            A[i][n] = A[i][n] - A[i][j] * hasil[j]
        hasil[i] = A[i][n]

    # Mengembalikan hasil berupa nilai x, y, z
    return hasil
```

- `def eliminasi_gauss(matriks):` → Mendefinisikan fungsi bernama `eliminasi_gauss` yang menerima matriks sebagai inputan
- `A = [baris[:] for baris in matriks]` → Membuat salinan matriks ke variabel baru bernama `A`, supaya matriks asli tidak ikut berubah saat proses perhitungan
- `n = len(A)` → Menghitung jumlah baris pada matriks, nilainya sama dengan jumlah variabel

```python
# Memanggil fungsi Eliminasi Gauss dan menampilkan hasil
x, y, z = eliminasi_gauss(matriks)

# Menampilkan hasil perhitungan
print("=== Hasil Penyelesaian SPL ===")
print(f"Nilai x = {x}")
print(f"Nilai y = {y}")
print(f"Nilai z = {z}")
```

Baris ini menjalankan fungsi `eliminasi_gauss` yang sudah dibuat sebelumnya. Data matriks yang kita definisikan di bagian pertama dikirim ke dalam fungsi untuk diproses. Hasil perhitungannya berupa 3 nilai langsung disimpan ke dalam variabel `x`, `y`, dan `z`.
